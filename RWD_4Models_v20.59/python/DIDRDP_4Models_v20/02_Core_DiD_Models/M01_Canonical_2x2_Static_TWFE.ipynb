## Formula
```
Y_it = alpha_i + gamma_t + beta*(D_i x Post_t) + e_it
```
**Source**: Goodman-Bacon (2021, J.Econometrics); clustering: Bertrand-Duflo-Mullainathan (2004, QJE)
